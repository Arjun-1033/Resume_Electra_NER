
# 3) Doccano Setup & Labeling

1. `docker run -d --name doccano -p 8080:8080 doccano/doccano`  
2. Create **Sequence Labeling** project with labels: `NAME, TITLE, PHONE, EMAIL, LINKEDIN, GITHUB, ORG, LOC, EDU, SKILL, CERT, PROJECT`  
3. Import `data/pre_annotated/preannotation_doccano.jsonl` (format: Doccano JSONL)  
4. Label/correct spans.  
5. Export JSONL to `data/annotated/export_doccano.jsonl`.
